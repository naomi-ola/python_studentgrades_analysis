```python
# Question 2: Create 'grade' column based on marks
def assign_grade(score):
    if score >= 85:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 60:
        return 'C'
    else:
        return 'D'

df['grade'] = df['score'].apply(assign_grade)
print(df)
print("\n" + "="*60 + "\n")
```

```python
import numpy as np

df['grade'] = np.where(df['score'] >= 85, 'A',
              np.where(df['score'] >= 70, 'B',
              np.where(df['score'] >= 60, 'C', 'D')))
df
```

```python
# Question 3: Sort by 'score' in descending order
print("Question 3: Sort DataFrame by 'score' in descending order")
print("-" * 60)
df_sorted = df.sort_values(by='score', ascending=False)
print(df_sorted)
```

```python
# Add grade column
def assign_grade(score):
    if score >= 85:
        return 'A'
    elif score >= 70:
        return 'B'
    elif score >= 60:
        return 'C'
    else:
        return 'D'

df['grade'] = df['score'].apply(assign_grade)

print("DataFrame with grade column:")
print(df)
print()

# Save to CSV file
df.to_csv('students_with_grades.csv', index=False)
print("✓ DataFrame saved to 'students_with_grades.csv'")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Set style for better-looking plots
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# Simple bar chart
plt.figure(figsize=(10, 6))
# Fill NaN values in 'name' column with an empty string for plotting
plot_names = df['name'].fillna('')
plt.bar(plot_names, df['score'], color='skyblue', edgecolor='black')
plt.xlabel('Student Name', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Scores by Student', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Count grades
grade_counts = df['grade'].value_counts()

plt.figure(figsize=(8, 8))
colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99']
plt.pie(grade_counts, labels=grade_counts.index, autopct='%1.1f%%',
        colors=colors, startangle=90, explode=[0.05, 0.05, 0.05, 0.05])
plt.title('Grade Distribution', fontsize=14, fontweight='bold')
plt.axis('equal')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
df.boxplot(column='score', by='class', grid=True, patch_artist=True)
plt.xlabel('Class', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Score Distribution by Class', fontsize=14, fontweight='bold')
plt.suptitle('')  # Remove default title
plt.tight_layout()
plt.show()
```

```python
# Create pivot table
pivot = df.pivot_table(values='score', index='class', columns='gender', aggfunc='mean')

plt.figure(figsize=(8, 6))
sns.heatmap(pivot, annot=True, fmt='.1f', cmap='YlOrRd',
            linewidths=1, cbar_kws={'label': 'Average Score'})
plt.title('Average Scores Heatmap (Class vs Gender)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
colors_map = {'A': 'green', 'B': 'blue', 'C': 'orange', 'D': 'red'}
colors = df['grade'].map(colors_map)

# Fill NaN values in 'name' column with an empty string for plotting
plot_names = df['name'].fillna('')
plt.scatter(plot_names, df['score'], c=colors, s=200, alpha=0.6, edgecolor='black')
plt.xlabel('Student Name', fontsize=12)
plt.ylabel('Score', fontsize=12)
plt.title('Student Scores (Colored by Grade)', fontsize=14, fontweight='bold')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df['score'], bins=10, color='lightgreen', edgecolor='black', alpha=0.7)
plt.xlabel('score', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.title('Distribution of Score', fontsize=14, fontweight='bold')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```
